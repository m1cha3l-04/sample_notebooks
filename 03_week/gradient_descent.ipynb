```python
import numpy as np
import math
```

```python
def gradient_descent(x,y):
    m_rec = b_rec = 0
    iter = 1000
    n = len(x)
    learning_rate = 0.08
    cost_previous = 0

    for i in range(iter):
        y_pred = m_rec * x + b_rec
        cost = (1/n) * sum([val ** 2 for val in (y - y_pred)])
        md = -(2/n) * sum (x*(y-y_pred))
        bd = -(2/n) * sum (y-y_pred)
        m_rec = m_rec - learning_rate * md
        b_rec = b_rec - learning_rate * bd
        #print("m {}, b {}, cost {} iteration {}".format(m_rec,b_rec,cost,i))
        if (math.isclose(cost, cost_previous, rel_tol=1e-20, abs_tol=1e-20)):
            print("break..")
            break
        cost_previous = cost
        print(cost_previous)

    
    #return m_rec, b_rec
        
```

```python
# 2x + 3
x = np.array([1,2,3,4,5])
y = np.array([5,7,9,11,13])
```

```python
gradient_descent(x,y)
```
