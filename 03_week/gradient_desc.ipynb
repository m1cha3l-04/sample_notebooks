```python
import numpy as np
```

```python
def gradient_desc(x,y):
    m_lat = b_lat = 0
    iter = 1000
    learning_rate = 0.08
    n = len(x)

    for i in range(iter):
        y_pred = m_lat * x + b_lat
        cost = (1/n) * sum([val**2 for val in (y-y_pred)])
        m_derivative = -(2/n) * sum(x*(y-y_pred))
        b_derivative = -(2/n) * sum(y-y_pred)

        m_lat = m_lat - learning_rate * m_derivative
        b_lat = b_lat - learning_rate * b_derivative

        print("m {}, b {}, cost {}, iteration {}".format(m_lat, b_lat, cost, i))
        
```

```python
x = np.array([1,2,3,4,5])
y = np.array([5,7,9,11,13])
```

```python
gradient_desc(x,y)
```
