```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
# Create dataframe
df = pd.read_csv('homeprices.csv')
df
```

```python
df.price
```

```python
#matplotlib inline
plt.xlabel('price')
plt.ylabel('area')
plt.scatter(df.area,df.price,color='red',marker='+')
```

```python
X = df.drop('price', axis='columns')
X
```

```python
y = df.price
y
```

```python
reg = linear_model.LinearRegression()
reg.fit(X,y)
```

```python
reg.predict([[3300]])
```

```python
reg.predict([[2800]])
```

```python
plt.xlabel('price')
plt.ylabel('area')
plt.scatter(df.area,df.price,color='red',marker='+')
plt.plot(X,reg.predict(X))
```

```python
# y = mx + b
m = reg.coef_
m
```

```python
b = reg.intercept_
b
```

```python
# 560821.91780822
y = m*2800 + b
y
```
